```python
import requests

API_KEY = "P0lOvnPLuEIrxopGDL27YGRmvgHr9T2gMEUYGPvG"
BASE_URL = "https://api.sportradar.com/tennis/trial/v3/en"

def test_endpoint(name, url):
    print(f"\n--- Testing: {name} ---")
    try:
        response = requests.get(url, timeout=10)
        print(f"Status Code: {response.status_code}")
        if response.status_code == 200:
            data = response.json()
            print(f"✅ SUCCESS - Keys in response: {list(data.keys())}")
        else:
            print(f"❌ FAILED - {response.text[:200]}")
    except Exception as e:
        print(f"❌ ERROR - {e}")

# Test all 3 endpoints
test_endpoint("Competitions",
    f"{BASE_URL}/competitions.json?api_key={API_KEY}")

test_endpoint("Complexes",
    f"{BASE_URL}/complexes.json?api_key={API_KEY}")

test_endpoint("Doubles Rankings",
    f"{BASE_URL}/double_competitors_rankings.json?api_key={API_KEY}")
```
